```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
```

```python
# Simple linear regression from Week 1
# Predicting house price based on size
size = np.array([1000, 1500, 2000, 2500, 3000])  # square feet
price = np.array([200, 250, 300, 350, 400])       # thousands

# Plot it
plt.scatter(size, price)
plt.xlabel('House Size (sq ft)')
plt.ylabel('Price ($1000s)')
plt.title('Simple Linear Regression - One Feature')
plt.show()

# Calculate mean
print(f"Average size: {np.mean(size)}")
print(f"Average price: {np.mean(price)}")
```

```python
# Multiple linear regression - predicting price with MULTIPLE features
# Features: size, bedrooms, age
houses = np.array([
    [1000, 2, 10],   # 1000 sq ft, 2 bedrooms, 10 years old
    [1500, 3, 5],    # 1500 sq ft, 3 bedrooms, 5 years old
    [2000, 3, 8],    
    [2500, 4, 3],
    [3000, 4, 15],
    [3500, 4, 8]
])

prices = np.array([200, 250, 300, 350, 400, 500])  # thousands

print("House features (size, bedrooms, age):")
print(houses)
print("\nPrices:")
print(prices)

# Calculate mean of each feature
print("\nMean of each feature:")
print(f"Average size: {np.mean(houses[:, 0])}")
print(f"Average bedrooms: {np.mean(houses[:, 1])}")
print(f"Average age: {np.mean(houses[:, 2])}")
```

```python
# Feature scaling - make all features similar scale
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
houses_scaled = scaler.fit_transform(houses)

print("Original features:")
print(houses)
print("\nScaled features (mean=0, std=1):")
print(houses_scaled)
```

Scaling the different features to be around the same value range is helpful for gradient descent. In looking at the gradient descent function, you can see that we are doing the derivative for each feature. That means that the step taken for a specific weight might be large while a step for another weight would be small. This will increase the amout of iterations needed to converge toward the solution
