```python
import numpy as np
import matplotlib.pyplot as plt

# Create simple data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Plot it
plt.plot(x, y, 'bo-')
plt.xlabel('Input')
plt.ylabel('Output')
plt.title('My First ML Plot - January 5, 2025')
plt.grid(True)
plt.show()

print("Environment setup successful! ✓")
```
